## Ejercicio 2: Conversor de temperaturas.

Existen diversas unidades de temperatura utilizadas en distintos contextos y regiones. Las más comunes son Celsius (°C), Fahrenheit (°F) y Kelvin (K). También existen otras unidades como Rankine (°Ra) y Réaumur (°Re). Selecciona al menos 2 conversores, de modo que al introducir una temperatura devuelva, como mínimo, dos conversiones.

```python
# Definimos las funciones de conversor para las 3 medidas disponibles:

def convertir_a_celsius(temperatura, unidad_entrada):
    if unidad_entrada == "farenheit":
        return (temperatura - 32) * 5 / 9
    elif unidad_entrada == "kelvin":
        return temperatura - 273.15
    elif unidad_entrada == "celsius":
        return temperatura
    else:
        return None

def convertir_a_farenheit(temperatura, unidad_entrada):
    if unidad_entrada == "celsius":
        return (temperatura * 9 / 5) + 32
    elif unidad_entrada == "kelvin":
        return (temperatura - 273.15) * 9 / 5 + 32
    elif unidad_entrada == "farenheit":
        return temperatura
    else:
        return None

def convertir_a_kelvin(temperatura, unidad_entrada):
    if unidad_entrada == "celsius":
        return temperatura + 273.15
    elif unidad_entrada == "farenheit":
        return (temperatura - 32) * 5 / 9 + 273.15
    elif unidad_entrada == "kelvin":
        return temperatura
    else:
        return None
    

# Pedimos los datos:   
def pedir_unidad_entrada():
    while True:
        unidad_entrada = input('Elige la unidad de medida de entrada (celsius, farenheit o kelvin): ').strip().lower()
        if unidad_entrada in ["celsius", "farenheit", "kelvin"]:
            print(f'Unidad de entrada seleccionada: {unidad_entrada}')
            return unidad_entrada
        else:
            print('Unidad no válida. Prueba de nuevo.')


def pedir_unidad_salida():
    while True:
        unidad_salida = input('Elige la unidad de medida que quieres obtener: celsius, farenheit o kelvin').strip().lower()
        if unidad_salida in ["celsius", "farenheit", "kelvin"]:
            print(f'Nueva unidad elegida: {unidad_salida}')
            return unidad_salida
        else:
            print('Unidad no válida. Prueba de nuevo.')


def pedir_temperatura():
    while True:
        temperatura = input('Introduce la temperatura:')
        try:
            temperatura = float(temperatura)
            return temperatura
        except ValueError:
            print('Por favor, introduce un número válido. Utiliza punto "." para los decimales.')

    
#Ejecutamos la funcion correspondiente segun los datos ingresados previamente:

def conversor_temperatura():
    print('Bienvenid@ al conversor de temperatura.')
   
    unidad_origen = pedir_unidad_entrada()
    unidad_destino = pedir_unidad_salida()
    temp_original = pedir_temperatura()
   
    print(f'\nTemperatura introducida: {temp_original} grados {unidad_origen}')

    if unidad_destino == 'celsius':
        resultado = convertir_a_celsius(temp_original, unidad_origen)
        simbolo = '°C'
    elif unidad_destino == 'farenheit':
        resultado = convertir_a_farenheit(temp_original, unidad_origen)
        simbolo = '°F'
    elif unidad_destino == 'kelvin':
        resultado = convertir_a_kelvin(temp_original, unidad_origen)
        simbolo = 'K'
    else:
        print('Unidad de destino no válida.')
        return

    print(f'\nResultado: {temp_original} grados {unidad_origen} = {round(resultado, 3)} {simbolo}')
```

## Ejercicio 3: Contador de palabras de un texto.

Escribe una función que dado un texto, muestre las veces que aparece cada palabra.

```python
import re

def limpiar_texto(texto):
    ''' Función que limpia el texto de signos y lo pasa a minúsculas'''
    
    texto_limpio = re.sub(r"[^a-zA-ZáéíóúÁÉÍÓÚàèìòùÀÈÌÒÙäëïöüÄËÏÖÜñÑ]", " ", texto).lower()
    return texto_limpio 


def contar_palabras(texto_limpio):
    ''' Función que cuenta las palabras de un texto'''

    palabras = texto_limpio.split()
    frecuencias = {}

    for palabra in palabras:
        if palabra not in frecuencias:
            frecuencias[palabra] = 1
        else:
            frecuencias[palabra] += 1
    return palabras, frecuencias


def longitud_media_palabras(palabras):
    ''' Funcion que calcula la media de longitud de las palabras.'''

    total_letras = sum(len(palabra) for palabra in palabras)
    media = round(total_letras / len(palabras), 1) if palabras else 0
    return media


def analizar_texto(texto):
    ''' Cuenta las palabras que hay en un texto, el numero de repeticiones de cada palabra y la longitud media de sus palabras'''

    texto_limpio = limpiar_texto(texto)
    palabras, frecuencias = contar_palabras(texto_limpio)
    media = longitud_media_palabras(palabras)

    print(f'Tu texto contiene {len(palabras)} palabras.')
    print(f'La cantidad de apariciones de cada palabra es: {frecuencias}.')
    print('La media de longitud de las palabras es de ', media, 'letras.')
```

```python
# Llamamos a la función final:
texto = input('Escribe tu texto.')
analizar_texto(texto)
```

## Ejercicio 4: Diccionario inverso.

Resulta que el cliente tiene una encuesta muy antigua que se almacena en un diccionario y los resultados los necesita **a la inversa**, es decir, intercambiados las claves y valores. Los valores y claves en el *diccionario original* son únicos; si éste no es el caso, la función debería imprimir un mensaje de advertencia.


```python
def diccionario_inverso(diccionario_original, verbose=False):
    ''' Función para invertir un diccionario con opción a saber si hay varios valores asociados a una clave'''
    
    diccionario_inverso = {}
    conflictos = []

    for clave, valor in diccionario_original.items():
        if valor not in diccionario_inverso:
            diccionario_inverso[valor] = clave
        else:
            conflictos.append(valor)
            
            # Si ya habia convertido el valor en lista, añado un nuevo valor a la lista    
            if isinstance(diccionario_inverso[valor], list):   
                diccionario_inverso[valor].append(clave)
            
            # Si el valor no era nua lista lo convierto en lista para poder añadir otros valores 
            else:
                diccionario_inverso[valor] = [diccionario_inverso[valor], clave]   
    
    if verbose and conflictos:
        print(f'Más de un valor asociado detectados en las claves: {set(conflictos)}')

    return diccionario_inverso
```

```python
diccionario_original = {'a': 1, 'b': 2, 'c': 3, 'd': 1, 'e': 2, 'f': 3, 'g': 1}
print(diccionario_inverso(diccionario_original,True))
```

## Exercici 1: Comptador i endreçador de paraules d'un text.

El client va quedar content amb el comptador de paraules, però ara vol llegir arxius TXT i que calculi la freqüència de cada paraula ordenades dins de les entrades habituals del diccionari segons la lletra amb la qual comencen, és a dir, les claus han d'anar de la A a la Z i dins de la A hem d'anar de la A la Z.

```python
def diccionario_recuento_palabras(ruta_archivo):
    import re

    with open(ruta_archivo, "r", encoding="utf-8") as archivo:
            texto = archivo.read()
    texto_limpio = re.sub(r"[^a-zA-ZáéíóúÁÉÍÓÚàèìòùÀÈÌÒÙäëïöüÄËÏÖÜñÑ]", " ", texto).lower() # Reemplazar por un espacio, cualquier cosa que NO sean los caracteres
                                                                                            # indicados (), y pasa a minusculas

    lista_palabras = texto_limpio.split()

    diccionario_palabras_frecuencia = {}           # Creo mi diccionario

    for palabra in lista_palabras:
        frecuencia = 0
        if palabra not in diccionario_palabras_frecuencia:
            frecuencia += 1
            diccionario_palabras_frecuencia[palabra] = frecuencia
        else:
            diccionario_palabras_frecuencia[palabra] += 1

    diccionario_palabras_frecuencia = sorted(diccionario_palabras_frecuencia.items()) 
                                                                        # Convierto el diccionario en una lista ordenada de tuplas (palabra,frecuencia). 
                                                                        # para añadirlos despues como valor al nuevo dicconario

    from collections import defaultdict

    diccionario_final = defaultdict(list) # Las claves se crean con el defaultdict, solo las que tengan valor asociado

    for palabra, frecuencia in sorted(diccionario_palabras_frecuencia):
            letra = palabra[0]                          # Se usa la priemra letra de la palabra para crear la clave del diccionario final 
            diccionario_final[letra].append((palabra, frecuencia))   # y añadirle el par (palabra, frecuencia) como valor

# Imprimir en columnas

    print("\nRecuento de palabras ordenadas alfabéticamente:\n")
    for letra, lista in sorted(diccionario_final.items()):
        print(f"\nLetra '{letra.upper()}':")
        
        for palabra, frecuencia in lista:
            print(f"-{palabra}: {frecuencia}")
```

## Ejercici 2: Conversió de tipus de dades.

El client rep una llista de dades i necessita generar dues llistes, la primera on estaran tots els elements que es van poder convertir en flotants i l'altra on estan els elements que no es van poder convertir.

```python
def conversor_tipos_datos(datos):     
    decimales = []
    otros = []

    def intentar_convertir(elemento):
        try:
            valor = float(elemento)         # Recorrer los datos y convertirlos a float y añadir a la lista de decimales
            decimales.append(valor)
        except (ValueError, TypeError):     # Si no se puede convertir, añadir a la lista 'otros'
            otros.append(elemento)

    for x in datos:                         # Leo en bucle los datos, y según el tipo, si son colecciones, entro en ellos y aplico la funcion intentar_convertir
        if isinstance(x, (list, tuple, set)):
            for elemento in x:
                intentar_convertir(elemento)

        elif isinstance(x, dict):
            for clave, valor in x.items():
                intentar_convertir(clave)
                intentar_convertir(valor)

        else:                               # En el resto de elementos que nos son colecciones aplico la función
            intentar_convertir(x)

    return decimales, otros                 # Me devuelve ambas listas completas
```

```python
datos_cliente = [ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']]

datos_separados = conversor_tipos_datos(datos_cliente)  # Llamamos a la función

decimales, otros = datos_separados  # Para extraer por separado cada lista

print(f'Los elementos decimales son: {decimales}.')
print(f'Los elementos que no pueden convertirse a decimal son: {otros}')
```

## Exercici 1: Generador de Contrasenyes

Explora el funcionament del mòdul random de la llibreria numpy.

En aquest punt, el client ha detectat un problema amb les contrasenyes que utilitzen els seus treballadors. Asdf1234, dates d’aniversari o similars. Per a solucionar-ho ens ha encarregat una funció de Python que generi contrasenyes més segures. La funció ha de dependre dels següents paràmetres:

- longitud (int): Longitud de la contrasenya
- majuscules (bool = True): Si hi ha d’aparèixer majúscules
- minuscules (bool = True): Si hi ha d’aparèixer minúscules
- numeros (bool = True): Si hi ha d’aparèixer números
- signes (bool = False): Si hi ha d’aparèixer caràcters especials (,-$? o similars)

Així doncs, si executem la funció de la següent manera:

crear_contrasenya(10, True, True, True, True)

Hauríem d’obtenir un output (que hem de poder guardar) de l’estil:

9Er,5Vn8P$

Assegura’t que es compleixin tots els criteris, i que aquestes contrasenyes siguin realment aleatòries.

(EXTRA) Explora com podríem fer que la funció copiés la contrasenya automàticament al porta papers de l’ordinador (com si l’haguéssim seleccionat i fet ctrl+copy).

```python
import numpy as np
from numpy import random
import string

def crear_contraseña(longitud=10, mayusculas=True, minusculas=True, numeros=True, signos=True):

    caracteres_validos = ''                                 # Genero una variable con todos los caracteres permitidos obteniendolos del modulo string
    if mayusculas:
        caracteres_validos += string.ascii_uppercase
    if minusculas:
        caracteres_validos += string.ascii_lowercase
    if numeros:
        caracteres_validos += string.digits
    if signos:
        caracteres_validos += string.punctuation
                                    
    lista_caracteres_validos = list(caracteres_validos)       # Convierto la variable en lista y de ella saldrán los caracteres para generar la contraseña

    if not lista_caracteres_validos:                # Para verificar que al menos uno de los caracteres es true (mayus, min...) y la lista no está vacá
        raise ValueError("Debes activar al menos un tipo de carácter (mayúsculas, minúsculas, números o signos).")
    
    while True:
        contraseña = np.random.choice(lista_caracteres_validos, size=longitud)    # Con np.random.choice cojo los elementos de la lista, con una longitud 
                                                                                # marcada por el argumento de la funcion antes definida. Me da una lista
        contraseña = ''.join(contraseña)                                    # Convierto esa lista en un string con .join() 
    
        if mayusculas and not any(c in string.ascii_uppercase for c in contraseña):
            continue
        if minusculas and not any(c in string.ascii_lowercase for c in contraseña):
            continue
        if numeros and not any(c in string.digits for c in contraseña):
            continue
        if signos and not any(c in string.punctuation for c in contraseña):
            continue

        return contraseña
        
nueva_contraseña = crear_contraseña()
print(f'La contraseña creada es: {nueva_contraseña}')


#para copiar al portapapeles:
import pyperclip

pyperclip.copy(nueva_contraseña)          # Copia al portapapeles
print(f"La contraseña '{nueva_contraseña}' se ha copiado al portapapeles.")


# para guardarlas en un archivo sin sobreescribir y con fecha y hora:
from datetime import datetime

fecha_hora_actual = datetime.now().strftime("%Y-%m-%d %H:%M:%S")        # Para obtener fecha y hora actual

with open("historial_contraseñas.txt", "a") as archivo:                 # Crea el archivo, lo abre, añade sin sobreescribir (modo 'a', de append)
    archivo.write(f"[{fecha_hora_actual}] Contraseña: {nueva_contraseña}\n")
    print(f"La contraseña '{nueva_contraseña}' se ha guardado en el archivo.")
```

## Exercici 2: Processament de dades simple

Una companya de feina ens ha demanat un favor, aprofitant que sap que estem aprenent a programar. Utilitza el fitxer ["historic_partits.txt"](historic_partits.txt)

Necessita un programa que retorni:

- El nombre total de gols que ha fet cada equip.
- El nom de l’equip més golejador.
- El nom de l’equip més golejat
- La classificació global (cada victòria: 3 pts, empat 1 pts, derrota 0 pts)

```python
estadisticas = {}

with open("historic_partits.txt", "r", encoding="utf-8") as archivo:

    lineas = archivo.readlines()    # Para leer linea a linea

    for linea in lineas:
        partes = linea.strip().split('\t')  # En cada linea, separa los 3 elementos y nos quedamos con los nombres de los equipos
        equipo_local = partes[0]
        resultado = partes[1]
        equipo_visitante = partes[2]

        if equipo_local not in estadisticas:    # Añadimos los nombres de los equipos al diccionario, como claves
            estadisticas[equipo_local] = {          # Y los valores serán otro subdiccionario con los campos que queremos calcular
  "ganados": 0,
  "empatados": 0,
  "perdidos": 0,
  "goles_favor": 0,
  "goles_contra": 0,
  "puntos": 0
}
        if equipo_visitante not in estadisticas:
            estadisticas[equipo_visitante] = {
  "ganados": 0,
  "empatados": 0,
  "perdidos": 0,
  "goles_favor": 0,
  "goles_contra": 0,
  "puntos": 0
}
        goles = resultado.strip().split('-')    # Separo los goles, 1ª cifra: local, 2ª cifra: visitante. Y paso a int para operar
        goles_local = int(goles[0])
        goles_visitante = int(goles[1])

        estadisticas[equipo_local]["goles_favor"] += goles_local    # Añado al diccionario los goles a favor y en contra correspondientes a cada equipo
        estadisticas[equipo_visitante]["goles_favor"] += goles_visitante
        estadisticas[equipo_local]["goles_contra"] += goles_visitante
        estadisticas[equipo_visitante]["goles_contra"] += goles_local

        if goles_local > goles_visitante:                       # En función de los goles de cada linea, adjudico los partidos ganados, perdidos,
                                                                # empatados y los adjudico a cada equipo
            estadisticas[equipo_local]['ganados'] += 1
            estadisticas[equipo_visitante]['perdidos'] +=1

        elif goles_local == goles_visitante:
            estadisticas[equipo_local]['empatados'] += 1
            estadisticas[equipo_visitante]['empatados'] +=1
        
        else:
            estadisticas[equipo_local]['perdidos'] += 1
            estadisticas[equipo_visitante]['ganados'] +=1

    for equipo in estadisticas:                             # Bucle para calcular los puntos  de cada equipo, que cuenta sus partidos ganados y empatados
        estadisticas[equipo]['puntos'] = (
            3 * estadisticas[equipo]['ganados'] +
            1 * estadisticas[equipo]['empatados']
        )

# Calculo los equipos con más goles y más goleados:

equipo_mas_goles = max(estadisticas, key=lambda equipo: estadisticas[equipo]["goles_favor"])
equipo_goleado = max(estadisticas, key=lambda equipo: estadisticas[equipo]["goles_contra"])

print(f'El equipo que ha marcado más goles es el {equipo_mas_goles}.')
print()
print(f'El equipo con más goleado es el {equipo_goleado}.')
print()
print("-" * 50)  # Línea de separación decorativa
print()


# Obtengo la clasificación ordenada por puntos:

clasificacion = sorted(estadisticas, key=lambda equipo: estadisticas[equipo]["puntos"], reverse=True)
print('La clasificación es:')
print()

for posicion, equipo in enumerate(clasificacion, start=1):
    print(f"{posicion}. {equipo} - {estadisticas[equipo]['puntos']} puntos")
```
